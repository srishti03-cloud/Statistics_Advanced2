Ques1.What is hypothesis testing in statistics?
Ans:Hypothesis testing is a structured way in statistics to make decisions using data. It helps you determine whether there’s enough evidence to support a claim or if an observed effect might just be due to random chance.
->Null Hypothesis (H₀): Assumes no effect or no difference
(e.g., “The new product does not improve customer satisfaction.”)
->Alternative Hypothesis (H₁ or Hₐ): What you want to test for
(e.g., “The new product improves customer satisfaction.”)

Ques2.What is the null hypothesis, and how does it differ from the alternative hypothesis?
Ans:The null hypothesis (\(H_{0}\)) is a default, testable assumption that no significant relationship, effect, or difference exists between variables, representing the "status quo". It is "null" because it usually suggests zero impact. It differs from the alternative hypothesis (\(H_{a}\) or \(H_{1}\)), which claims a significant effect or difference exists.

Ques3.Explain the significance level in hypothesis testing and its role in deciding the outcome of a test.
Ans:The significance level (denoted as \(\alpha \) or alpha) is the pre-set threshold probability of rejecting a true null hypothesis, often called a Type I error or "false positive". Common levels are \(0.05\) (5%), \(0.01\) (1%), or \(0.10\) (10%). It dictates the outcome by comparing \(\alpha \) to the p-value: if the p-value \(\le \alpha \), results are statistically significant, and we reject the null hypothesis.

Role in Deciding Test Outcomes
The significance level works with the p-value to determine whether to reject the null hypothesis:
If \(P\text{-value} \le \alpha\): Reject the null hypothesis (\(H_{0}\)). The results are statistically significant.
If \(P\text{-value} > \alpha\): Fail to reject the null hypothesis (\(H_{0}\)). The results are not statistically significant.

Ques4.What are Type I and Type II errors? Give examples of each.
Ans:Type I errors (false positives) occur when a true null hypothesis is incorrectly rejected, often described as finding a "false alarm" or an effect that isn't real (\(H_{0}\) is true, but rejected). Type II errors (false negatives) occur when a false null hypothesis is failed to be rejected, missing a real effect or "missing an alarm" (\(H_{0}\) is false, but accepted). These errors, common in statistics, legal, and medical fields, are often inversely related: lowering the probability of a Type I error (\(\alpha \)) often increases the probability of a Type II error (\(\beta \)).

Type I Error (False Positive) Examples
Medical Testing: A patient is diagnosed with a disease they do not have.Jury Trial: A jury finds a defendant guilty (rejecting innocence) when they are actually innocent.

Type II Error (False Negative) Examples
Medical Testing: A test incorrectly indicates a person is free from a disease when they actually have it.
Jury Trial: A jury finds a defendant not guilty (failing to reject innocence) when they are actually guilty.

Ques5.What is the difference between a Z-test and a T-test? Explain when to use each.
Ans:The primary difference between a Z-test and a T-test is whether the population standard deviation (\(\sigma \)) is known. Use a Z-test for large samples (\(n \ge 30\)) or when \(\sigma \) is known (using the normal distribution). Use a T-test for small samples (\(n < 30\)) or when \(\sigma \) is unknown, relying on sample standard deviation (\(s\)) and the \(t\)-distribution.

Use a Z-test when:
1.The population standard deviation (\(\sigma \)) is known.
2.The sample size is large (\(n > 30\)), even if \(\sigma \) is unknown, due to the Central Limit Theorem.
3.Testing proportions (e.g., A/B testing conversion rates).
Use a T-test when:
1.The population standard deviation (\(\sigma \)) is unknown (most common scenario).
2.The sample size is small (\(n < 30\)).
3.Data is approximately normally distributed.

```python
"""Implement hypothesis testing using Z-statistics for a sample dataset in
Python. Show the Python code and interpret the results.
sample_data = [49.1, 50.2, 51.0, 48.7, 50.5, 49.8, 50.3, 50.7, 50.2, 49.6,
50.1, 49.9, 50.8, 50.4, 48.9, 50.6, 50.0, 49.7, 50.2, 49.5,
50.1, 50.3, 50.4, 50.5, 50.0, 50.7, 49.3, 49.8, 50.2, 50.9,
50.3, 50.4, 50.0, 49.7, 50.5, 49.9]
(Include your Python code and output in the code box below.)"""


# Import libraries
import numpy as np
from scipy.stats import norm

# Given sample data
sample_data = [49.1, 50.2, 51.0, 48.7, 50.5, 49.8, 50.3, 50.7, 50.2, 49.6,
               50.1, 49.9, 50.8, 50.4, 48.9, 50.6, 50.0, 49.7, 50.2, 49.5,
               50.1, 50.3, 50.4, 50.5, 50.0, 50.7, 49.3, 49.8, 50.2, 50.9,
               50.3, 50.4, 50.0, 49.7, 50.5, 49.9]

# Convert to numpy array
data = np.array(sample_data)

# Sample statistics
mean_sample = np.mean(data)
std_sample = np.std(data, ddof=1)
n = len(data)

# Hypothesized population mean
mu = 50

# Compute Z-statistic
z = (mean_sample - mu) / (std_sample / np.sqrt(n))

# Two-tailed p-value
p_value = 2 * (1 - norm.cdf(abs(z)))

# Print results
print("Sample Mean:", round(mean_sample, 4))
print("Sample Std Dev:", round(std_sample, 4))
print("Z-statistic:", round(z, 4))
print("P-value:", round(p_value, 4))

# Decision at alpha = 0.05
alpha = 0.05
if p_value < alpha:
    print("Reject the null hypothesis (H0)")
else:
    print("Fail to reject the null hypothesis (H0)")

```

```python
""" Write a Python script to simulate data from a normal distribution and
calculate the 95% confidence interval for its mean. Plot the data using Matplotlib.
(Include your Python code and output in the code box below.)"""


# Import libraries
import numpy as np
import matplotlib.pyplot as plt
from scipy import stats

# Step 1: Simulate data from a normal distribution
np.random.seed(42)
data = np.random.normal(loc=50, scale=5, size=100)  # mean=50, std=5, n=100

# Step 2: Compute sample statistics
mean = np.mean(data)
std_err = stats.sem(data)  # standard error

# Step 3: Compute 95% confidence interval
confidence = 0.95
df = len(data) - 1
ci = stats.t.interval(confidence, df, loc=mean, scale=std_err)

# Step 4: Print results
print("Sample Mean:", round(mean, 4))
print("95% Confidence Interval:", (round(ci[0], 4), round(ci[1], 4)))

# Step 5: Plot histogram
plt.hist(data, bins=15, edgecolor='black', alpha=0.7)

# Plot mean and confidence interval lines
plt.axvline(mean, color='red', linestyle='dashed', linewidth=2, label='Mean')
plt.axvline(ci[0], color='green', linestyle='dashed', linewidth=2, label='Lower CI')
plt.axvline(ci[1], color='green', linestyle='dashed', linewidth=2, label='Upper CI')

# Labels and title
plt.title("Normal Distribution with 95% Confidence Interval")
plt.xlabel("Values")
plt.ylabel("Frequency")
plt.legend()

# Show plot
plt.show()
```

```python
"""Write a Python function to calculate the Z-scores from a dataset and
visualize the standardized data using a histogram. Explain what the Z-scores represent
in terms of standard deviations from the mean.
(Include your Python code and output in the code box below.)"""


# Import libraries
import numpy as np
import matplotlib.pyplot as plt

# Function to calculate Z-scores
def calculate_z_scores(data):
    data = np.array(data)
    mean = np.mean(data)
    std = np.std(data, ddof=1)
    z_scores = (data - mean) / std
    return z_scores, mean, std

# Example dataset
data = [10, 12, 13, 12, 11, 14, 13, 15, 16, 12,
        11, 13, 14, 15, 16, 17, 18, 19, 12, 11]

# Calculate Z-scores
z_scores, mean, std = calculate_z_scores(data)

# Print results
print("Mean:", round(mean, 4))
print("Standard Deviation:", round(std, 4))
print("First 10 Z-scores:", np.round(z_scores[:10], 4))

# Plot histogram of Z-scores
plt.hist(z_scores, bins=10, edgecolor='black', alpha=0.7)
plt.title("Histogram of Z-scores (Standardized Data)")
plt.xlabel("Z-score")
plt.ylabel("Frequency")

# Show plot
plt.show()

```
